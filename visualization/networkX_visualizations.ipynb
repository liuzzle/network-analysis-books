### Imports

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
```

### Load the Relationship Data

Book 1 (Alice's Adventures in Wonderland)

```python
relationship_df_b1 = pd.read_csv('book1_relationship_data.csv')
```

```python
# Create a graph from the relationship DataFrame
G = nx.from_pandas_edgelist(relationship_df_b1, 
                            source='source', 
                            target='target', 
                            edge_attr='count', 
                            create_using=nx.Graph())

# Draw the graph
plt.figure(figsize=(10, 10))
pos = nx.spring_layout(G)  # Layout for the nodes
nx.draw(G, pos, with_labels=True, node_color='skyblue', edge_cmap=plt.cm.Blues)

# Draw edge labels with normalized count
edge_labels = nx.get_edge_attributes(G, 'normalized_count')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.show()
```
